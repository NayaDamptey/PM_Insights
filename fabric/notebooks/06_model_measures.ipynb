```python
from sempy_labs.tom import connect_semantic_model

added, updated, removed = 0, 0, 0
with connect_semantic_model(dataset=semantic_model, workspace=workspace, readonly=False) as tom:
    # Measures that were renamed: the old names would otherwise stay in the model.
    for table_name, measure_name in MODEL_MEASURES["removed"]:
        old = tom.model.Tables[table_name].Measures.Find(measure_name)
        if old is not None:
            tom.remove_object(old)
            removed += 1

    for m in MODEL_MEASURES["measures"]:
        table = tom.model.Tables[m["table"]]
        measure = table.Measures.Find(m["name"])
        if measure is None:
            tom.add_measure(
                table_name=m["table"],
                measure_name=m["name"],
                expression=m["expression"],
                format_string=m["format"] or None,
                display_folder=m["folder"],
                description=m["description"] or None,
            )
            measure = table.Measures.Find(m["name"])
            added += 1
        else:
            measure.Expression = m["expression"]
            measure.FormatString = m["format"]
            measure.DisplayFolder = m["folder"]
            measure.Description = m["description"]
            updated += 1
        if m["data_category"]:
            measure.DataCategory = m["data_category"]

print(f"{semantic_model}: {added} measures added, {updated} updated, {removed} old ones removed")
```
